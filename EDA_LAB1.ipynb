```python
import pandas as pd

df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df["Episodes"] = pd.to_numeric(df["Episodes"], errors="coerce")
df["Rank"] = pd.to_numeric(df["Rank"], errors="coerce")
df["Scored By"] = pd.to_numeric(df["Scored By"], errors="coerce")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(df["Score"].dropna(), bins=20, edgecolor="black")

plt.title("Distribution of Anime Scores")
plt.xlabel("Score")
plt.ylabel("Number of Anime")

plt.show()
```

```python
df["Type"].value_counts()
```

```python
import matplotlib.pyplot as plt

type_counts = df["Type"].value_counts()

plt.figure(figsize=(8, 5))
type_counts.plot(kind="bar", color="steelblue", edgecolor="black")

plt.title("Number of Anime by Type")
plt.xlabel("Anime Type")
plt.ylabel("Number of Anime")
plt.xticks(rotation=45)

plt.show()
```

```python
# Separate multiple genres and count each genre
genre_counts = df["Genres"].str.split(", ").explode().value_counts()

genre_counts
plt.figure(figsize=(10, 6))

genre_counts.head(10).plot(
    kind="bar",
    color="coral",
    edgecolor="black"
)

plt.title("Top 10 Most Common Anime Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Anime")
plt.xticks(rotation=45)

plt.show()

```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["Score"].dropna())

plt.title("Box Plot of Anime Scores")
plt.ylabel("Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Score"], df["Members"], alpha=0.3)

plt.title("Anime Score vs Members")
plt.xlabel("Score")
plt.ylabel("Members")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Score"], df["Members"], alpha=0.3)

plt.yscale("log")

plt.title("Anime Score vs Members")
plt.xlabel("Score")
plt.ylabel("Members (log scale)")

plt.show()
```

```python
correlation = df["Score"].corr(df["Members"])

print("Correlation between Score and Members:", correlation)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = [
    "Score",
    "Episodes",
    "Rank",
    "Popularity",
    "Favorites",
    "Scored By",
    "Members"
]

correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Anime Variables")
plt.show()
```

```python
# Extract the starting year from the Aired column
df["Year"] = pd.to_datetime(
    df["Aired"].str.extract(r"(\d{4})")[0],
    errors="coerce"
).dt.year
year_counts = df["Year"].value_counts().sort_index()
plt.figure(figsize=(12, 5))

plt.plot(year_counts.index, year_counts.values)

plt.title("Number of Anime Released by Year")
plt.xlabel("Year")
plt.ylabel("Number of Anime")

plt.grid(True, alpha=0.3)
plt.show()
```
